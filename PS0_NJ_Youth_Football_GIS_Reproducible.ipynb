### 5.2 Importing New Jersey Geographic Data

The first geographic dataset I will load contains the boundaries of New Jersey's 21 counties.

I will retrieve this information from the New Jersey Office of GIS, which provides publicly available mapping data.

The information will be downloaded using Python, converted into a shapefile, and loaded with GeoPandas.

A shapefile is a common type of geographic data file that allows computers to display locations and boundaries on a map.

These county boundaries will help show where youth football organizations are located throughout New Jersey.

**Data source:** New Jersey Office of Information Technology, Office of GIS (NJOGIS).

https://services2.arcgis.com/XVOqAjTOJ5P6ngMu/arcgis/rest/services/NJ_Counties_3424/FeatureServer/0

```python

# -------------------------------------------------------
# PS0 - NEW JERSEY YOUTH FOOTBALL GIS
# SECTION 5.2A - IMPORT NJ COUNTY BOUNDARIES
# -------------------------------------------------------

import requests
import geopandas as gpd
from pathlib import Path

# Official NJ Office of GIS online data source
county_url = (
    "https://services2.arcgis.com/"
    "XVOqAjTOJ5P6ngMu/arcgis/rest/services/"
    "NJ_Counties_3424/FeatureServer/0/query"
)

# Request all county boundaries in geographic coordinates
county_params = {
    "where": "1=1",
    "outFields": "*",
    "returnGeometry": "true",
    "outSR": 4326,
    "f": "geojson"
}

response = requests.get(
    county_url,
    params=county_params,
    timeout=90
)

response.raise_for_status()
county_data = response.json()

assert county_data.get("type") == "FeatureCollection", (
    f"Unexpected response: {str(county_data)[:300]}"
)

# Convert the downloaded information to a GeoDataFrame
county_source = gpd.GeoDataFrame.from_features(
    county_data["features"],
    crs="EPSG:4326"
)

# Verify all 21 NJ counties were retrieved
assert len(county_source) == 21, (
    f"Expected 21 counties; found {len(county_source)}"
)

assert "GNIS_NAME" in county_source.columns

# Keep the county names and geographic boundaries
county_clean = gpd.GeoDataFrame(
    {
        "COUNTY": county_source["GNIS_NAME"].astype(str)
    },
    geometry=county_source.geometry,
    crs=county_source.crs
)

assert county_clean["COUNTY"].nunique() == 21

# Create a temporary PS0 shapefile in Colab
county_folder = Path("/content/ps0_data/nj_counties")
county_folder.mkdir(parents=True, exist_ok=True)

county_shapefile = county_folder / "nj_counties.shp"

county_clean.to_file(
    county_shapefile,
    driver="ESRI Shapefile",
    index=False
)

# Read the shapefile back into Python
nj_counties = gpd.read_file(county_shapefile)

# Final data checks
assert len(nj_counties) == 21
assert nj_counties.crs is not None
assert nj_counties.geometry.notna().all()

# Display results
print("PS0 - NJ COUNTY BOUNDARY IMPORT")
print("-" * 45)

print("County records:", len(nj_counties))
print("Coordinate system:", nj_counties.crs)
print("Geometry types:", nj_counties.geom_type.unique())
print("Shapefile:", county_shapefile)

print("-" * 45)
print("NJ county shapefile loaded successfully.")

display(nj_counties[["COUNTY"]].head())
```

### 5.2B Importing New Jersey Municipal Boundaries

The next dataset contains the geographic boundaries of New Jersey's 564 municipalities.

Municipalities are local communities such as cities, boroughs, and townships. Each municipality is located within one of New Jersey's 21 counties.

I will use geographic information from the New Jersey Office of GIS to identify these municipal boundaries.

The data will be downloaded using Python, converted into a shapefile, and loaded using GeoPandas.

These boundaries will provide more detailed geographic information for understanding where community youth football organizations are located.

This information may also be useful for future research involving local governments, regional planning, and the proposed NJYASA statewide structure.

**Data source:** New Jersey Office of Information Technology, Office of GIS (NJOGIS).

https://services2.arcgis.com/XVOqAjTOJ5P6ngMu/ArcGIS/rest/services/NJ_Municipal_Boundaries_3424/FeatureServer/0

```python

# -------------------------------------------------------
# PS0 - NEW JERSEY YOUTH FOOTBALL GIS
# SECTION 5.2B - IMPORT NJ MUNICIPAL BOUNDARIES
# -------------------------------------------------------

import requests
import geopandas as gpd
from pathlib import Path

# Official NJ Office of GIS data source
municipality_url = (
    "https://services2.arcgis.com/"
    "XVOqAjTOJ5P6ngMu/ArcGIS/rest/services/"
    "NJ_Municipal_Boundaries_3424/FeatureServer/0/query"
)

# Request all municipal geographic boundaries
municipality_params = {
    "where": "1=1",
    "outFields": "MUN_CODE,NAME,COUNTY",
    "returnGeometry": "true",
    "outSR": 4326,
    "f": "geojson"
}

response = requests.get(
    municipality_url,
    params=municipality_params,
    timeout=120
)

response.raise_for_status()

municipality_data = response.json()

assert municipality_data.get("type") == "FeatureCollection", (
    f"Unexpected response: {str(municipality_data)[:300]}"
)

# Convert downloaded data into geographic format
municipality_source = gpd.GeoDataFrame.from_features(
    municipality_data["features"],
    crs="EPSG:4326"
)

# Verify expected municipal records
assert len(municipality_source) == 564, (
    f"Expected 564 municipalities; "
    f"found {len(municipality_source)}"
)

required_columns = ["MUN_CODE", "NAME", "COUNTY"]

assert all(
    col in municipality_source.columns
    for col in required_columns
), "Required municipal data fields are missing."

# Keep the municipality name, county, code, and geometry
municipality_clean = gpd.GeoDataFrame(
    {
        "MUN_CODE": municipality_source["MUN_CODE"],
        "MUN_NAME": municipality_source["NAME"],
        "COUNTY": municipality_source["COUNTY"]
    },
    geometry=municipality_source.geometry,
    crs=municipality_source.crs
)

assert municipality_clean.geometry.notna().all()

# Create a temporary municipal shapefile
municipality_folder = Path(
    "/content/ps0_data/nj_municipalities"
)

municipality_folder.mkdir(
    parents=True,
    exist_ok=True
)

municipality_shapefile = (
    municipality_folder / "nj_municipalities.shp"
)

municipality_clean.to_file(
    municipality_shapefile,
    driver="ESRI Shapefile",
    index=False
)

# Read the shapefile back into Python
nj_municipalities = gpd.read_file(
    municipality_shapefile
)

# Final checks
assert len(nj_municipalities) == 564
assert nj_municipalities.crs.to_epsg() == 4326
assert nj_municipalities.geometry.notna().all()

# Display results
print("PS0 - NJ MUNICIPAL BOUNDARY IMPORT")
print("-" * 45)

print(
    "Municipal records:",
    len(nj_municipalities)
)

print(
    "Coordinate system:",
    nj_municipalities.crs
)

print(
    "Geometry types:",
    nj_municipalities.geom_type.unique()
)

print(
    "Shapefile:",
    municipality_shapefile
)

print("-" * 45)
print("NJ municipal shapefile loaded successfully.")

display(
    nj_municipalities[
        ["MUN_NAME", "COUNTY", "MUN_CODE"]
    ].head()
)
```

### 5.2C Importing the New Jersey Youth Football Research Dataset

For this part of the project, I will use the youth football research dataset that I previously developed and published on GitHub.

The dataset contains 265 documented community youth football organizations in New Jersey. It includes organization names, geographic coordinates, and information about whether each organization is included in the geographic analysis.

Of the 265 organizations, 264 are currently included in the map. One organization, P181, is excluded pending further organization verification.

I will use Python to download this information directly from GitHub and create geographic points using the recorded latitude and longitude coordinates.

This allows my professor, classmates, and other readers to reproduce the map without needing access to my personal Google Drive.

The published file is based on research and geographic verification completed in a separate data preparation project. It is a prepared research dataset, rather than the original information collected from leagues and organizations.

```python

# -------------------------------------------------------
# PS0 - NEW JERSEY YOUTH FOOTBALL GIS
# SECTION 5.2C - IMPORT PUBLIC RESEARCH DATA
# SOURCE: GITHUB
# NO GOOGLE DRIVE REQUIRED
# -------------------------------------------------------

import pandas as pd
import geopandas as gpd
import requests
from io import BytesIO

# Public research data published on GitHub
football_data_url_ps0 = (
    "https://raw.githubusercontent.com/"
    "Dunk0725/GIS-26/"
    "NJ-Youth-Football-GIS/"
    "PS0_NJ_Youth_Football_Locations_2026.csv"
)

# 1. Download the public research dataset
response = requests.get(
    football_data_url_ps0,
    timeout=60
)

response.raise_for_status()

public_data_ps0 = pd.read_csv(
    BytesIO(response.content),
    dtype={"Program_ID": "string"}
)

# 2. Verify required research fields
required_columns = [
    "Program_ID",
    "Canonical_Program_Name",
    "Latitude",
    "Longitude",
    "Geographic_QA_Category",
    "Analysis_Eligibility"
]

assert list(public_data_ps0.columns) == required_columns

public_data_ps0["Program_ID"] = (
    public_data_ps0["Program_ID"].str.strip()
)

assert len(public_data_ps0) == 265
assert public_data_ps0["Program_ID"].notna().all()
assert public_data_ps0["Program_ID"].is_unique

# 3. Identify organizations eligible for mapping
eligible_ps0 = public_data_ps0[
    "Analysis_Eligibility"
].eq(
    "Included - subject to documented QA limitations"
)

excluded_ps0 = public_data_ps0[
    "Analysis_Eligibility"
].eq(
    "Excluded pending organization verification"
)

assert eligible_ps0.sum() == 264
assert excluded_ps0.sum() == 1
assert (eligible_ps0 | excluded_ps0).all()

assert public_data_ps0.loc[
    excluded_ps0, "Program_ID"
].tolist() == ["P181"]

# 4. Select eligible organizations
football_rows_ps0 = public_data_ps0.loc[
    eligible_ps0
].copy()

football_rows_ps0["Latitude"] = pd.to_numeric(
    football_rows_ps0["Latitude"],
    errors="raise"
)

football_rows_ps0["Longitude"] = pd.to_numeric(
    football_rows_ps0["Longitude"],
    errors="raise"
)

assert football_rows_ps0[
    ["Latitude", "Longitude"]
].notna().all().all()

# 5. Create geographic points from coordinates
football_points_ps0 = gpd.GeoDataFrame(
    football_rows_ps0,
    geometry=gpd.points_from_xy(
        football_rows_ps0["Longitude"],
        football_rows_ps0["Latitude"]
    ),
    crs="EPSG:4326"
).sort_values(
    "Program_ID"
).reset_index(drop=True)

# 6. Final verification
assert len(football_points_ps0) == 264
assert football_points_ps0.geometry.is_valid.all()
assert football_points_ps0.geom_type.eq("Point").all()
assert football_points_ps0.crs.to_epsg() == 4326

print("PS0 - PUBLIC RESEARCH DATA IMPORT")
print("-" * 50)

print("GitHub HTTP status:", response.status_code)
print("Total organizations:", len(public_data_ps0))
print("Organizations mapped:", len(football_points_ps0))
print("Excluded organization: P181")
print("Coordinate system:", football_points_ps0.crs)

print()
print("FIRST FIVE ORGANIZATIONS")

display(
    football_points_ps0[
        [
            "Program_ID",
            "Canonical_Program_Name",
            "Latitude",
            "Longitude"
        ]
    ].head()
)

print()
print("SECTION 5.2C COMPLETE")
print("Public GIS points created successfully.")
print("No Google Drive access required.")
print("No original research files modified.")
```

Some organizations may share facilities, meaning that more than one organization can appear at the same location on the map.

```python

# -------------------------------------------------------
# PS0 - NEW JERSEY YOUTH FOOTBALL GIS
# STEP 6 - GEOGRAPHIC QA AND FIRST DRAFT MAP
# READ-ONLY - NO RESEARCH FILES MODIFIED
# -------------------------------------------------------

import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd

# -------------------------------------------------------
# 1. Verify the three previously loaded datasets
# -------------------------------------------------------

assert len(nj_counties) == 21
assert len(nj_municipalities) == 564
assert len(football_points_ps0) == 264

assert football_points_ps0["Program_ID"].is_unique

assert nj_counties.crs is not None
assert nj_municipalities.crs is not None
assert football_points_ps0.crs is not None

assert football_points_ps0.geometry.notna().all()
assert football_points_ps0.geometry.is_valid.all()
assert football_points_ps0.geom_type.eq("Point").all()

# -------------------------------------------------------
# 2. Prepare a common map coordinate system
# EPSG:3424 - New Jersey State Plane
# -------------------------------------------------------

county_map_ps0 = nj_counties.to_crs(
    "EPSG:3424"
)

municipality_map_ps0 = nj_municipalities.to_crs(
    "EPSG:3424"
)

football_map_ps0 = football_points_ps0.to_crs(
    "EPSG:3424"
)

# -------------------------------------------------------
# 3. Verify points fall inside New Jersey
# -------------------------------------------------------

state_boundary_ps0 = (
    county_map_ps0.geometry.union_all()
)

inside_nj_ps0 = football_map_ps0.geometry.apply(
    state_boundary_ps0.covers
)

print("PS0 - GEOGRAPHIC QUALITY CHECK")
print("-" * 50)

print("Counties:", len(county_map_ps0))
print("Municipalities:", len(municipality_map_ps0))
print("Football organizations:", len(football_map_ps0))

print("Map coordinate system:", football_map_ps0.crs)

print(
    "Locations within NJ boundaries:",
    int(inside_nj_ps0.sum())
)

print(
    "Locations requiring boundary review:",
    int((~inside_nj_ps0).sum())
)

# Stop and report any location outside the state
if not inside_nj_ps0.all():

    print("\nLOCATIONS REQUIRING REVIEW")

    display(
        football_map_ps0.loc[
            ~inside_nj_ps0,
            [
                "Program_ID",
                "Canonical_Program_Name",
                "Latitude",
                "Longitude"
            ]
        ]
    )

    raise ValueError(
        "Geographic boundary exceptions require "
        "review before generating the map."
    )

print("\nGeographic checks passed.")

# -------------------------------------------------------
# 4. Create the first draft GIS map
# -------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(9, 11)
)

# New Jersey counties as the geographic background
county_map_ps0.plot(
    ax=ax,
    facecolor="#F2F2F2",
    edgecolor="#666666",
    linewidth=0.8
)

# Documented home-field locations
football_map_ps0.plot(
    ax=ax,
    color="#A32938",
    markersize=14,
    alpha=0.8,
    label="Youth football organizations"
)

# Map title
ax.set_title(
    "Community Youth Football Organizations\n"
    "in New Jersey (2026)",
    fontsize=15,
    pad=16
)

# Legend
ax.legend(
    loc="lower left",
    fontsize=9,
    frameon=False
)

# Hide geographic coordinate axes
ax.set_axis_off()

# Clarifying note
fig.text(
    0.5,
    0.035,
    "264 documented organizations shown by home-field location.\n"
    "One organization excluded pending verification.",
    ha="center",
    fontsize=9
)

plt.tight_layout(rect=[0, 0.07, 1, 1])
plt.show()

# -------------------------------------------------------
# 5. Completion summary
# -------------------------------------------------------

print("\nSTEP 6 COMPLETE")
print("Geographic layers verified.")
print("First draft map displayed.")
print("No research files modified.")
print("No files written or overwritten.")
```
